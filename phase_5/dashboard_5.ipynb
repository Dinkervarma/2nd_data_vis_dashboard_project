```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("corporate_dataset.csv");
```

```python
df.head()
```

```python
corr_columns = [
    "ai_adoption_level",
    "ai_investment_usd",
    "automation_rate",
    "cost_savings",
    "revenue_impact",
    "productivity_gain",
    "employee_ai_training_hours",
    "ai_maturity_score",
    "deployment_count"
]

corr = df[corr_columns].corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title("Correlation Matrix Heatmap", fontsize=16)

plt.xticks(rotation=45, ha="right")

# Display heatmap
plt.tight_layout()
plt.show()

print("Observations")
print("----------------------------")
print("1. Positive values indicate a positive relationship.")
print("2. Negative values indicate an inverse relationship.")
print("3. Values closer to 1 or -1 show stronger correlation.")
print("4. Values near 0 indicate weak or no linear relationship.")
```

```python
#to find the averages of KPI(Key Performance Insights) and Predictive Performance Metrics
target_columns = [
    'ai_adoption_level', 
    'productivity_gain', 
    'revenue_impact', 
    'ai_investment_usd', 
    'employee_ai_training_hours'
]

averages = df[target_columns].mean().round(2)

print(averages)
```
